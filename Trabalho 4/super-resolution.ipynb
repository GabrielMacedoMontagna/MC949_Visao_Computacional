Aqui optaremos por usar o pipeline disponível do Stable Diffusion 2.0 - Upscaler 4×, já que ele foi treinado para aumentar resolução de imagens por fator 4. Esse modelo de difusão condiciona a geração de detalhes de alta frequência à imagem de baixa resolução fornecida, podendo também usar um prompt de texto para guiar detalhes.

```python
import torch
from diffusers import StableDiffusionUpscalePipeline

model_id = "stabilityai/stable-diffusion-x4-upscaler"  # modelo 4x upscaler do Stable Diffusion 2.0
# Carrega o pipeline com meio de precisão (fp16) para acelerar e economizar memória (requer GPU compatível)
pipeline = StableDiffusionUpscalePipeline.from_pretrained(
    model_id, revision="fp16", torch_dtype=torch.float16
)
pipeline = pipeline.to("cuda")  # move para GPU; use "cpu" se não tiver GPU (neste caso remova fp16)
```
